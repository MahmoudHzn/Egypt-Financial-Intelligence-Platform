#LSTM

```python
features = df3.drop(columns=["Gold_Futures", "Date"],axis=1)

target = df3["Gold_Futures"]
```

```python
features
```

```python
X_data = features.values
y_data = target.values
```

```python
from sklearn.preprocessing import MinMaxScaler

x_scaler = MinMaxScaler()
y_scaler = MinMaxScaler()
```

```python
X_scaled = x_scaler.fit_transform(features)
```

```python
y_scaled = y_scaler.fit_transform(
    target.values.reshape(-1,1)
)
```

```python
sequence_length = 30

X = []
y = []

for i in range(sequence_length, len(df3)):

    X.append(
        X_scaled[i-sequence_length:i]
    )

    y.append(
        y_scaled[i]
    )

X = np.array(X)
y = np.array(y)
```

```python
print(X.shape)
print(y.shape)
```

```python
split = int(len(X) * 0.8)

X_train = X[:split]
X_test  = X[split:]

y_train = y[:split]
y_test  = y[split:]
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
```

```python
model = Sequential([

    LSTM(
        64,
        input_shape=(X.shape[1], X.shape[2])
    ),

    Dropout(0.2),

    Dense(32, activation="relu"),

    Dense(1)

])
```

```python
model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)
```

```python
history = model.fit(

    X_train,
    y_train,

    epochs=20,

    batch_size=32,

    validation_data=(X_test, y_test)

)
```

```python
predictions = model.predict(X_test)
```
